```python
from langdetect import detect

# Get text from user
text = input("Enter a sentence: ")

# Detect language
language = detect(text)

# Language names
languages = {
    "en": "English",
    "ta": "Tamil",
    "hi": "Hindi",
    "te": "Telugu",
    "ml": "Malayalam",
    "kn": "Kannada",
    "fr": "French",
    "de": "German",
    "es": "Spanish",
    "it": "Italian",
    "ar": "Arabic",
    "ja": "Japanese",
    "ko": "Korean",
    "zh-cn": "Chinese"
}

# Display result
print("\nLANGUAGE DETECTION")
print("----------------------")

print("Input Text:", text)
print("Language Code:", language)
print("Detected Language:", languages.get(language, "Unknown Language"))
```
